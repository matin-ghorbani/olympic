# clfkit demo

Quick smoke-test for the classification benchmark pipeline.

Give it a **preprocessed** DataFrame; it tunes several models for **accuracy** and prints a leaderboard plus the winning hyperparameters.

```python
import sys
from pathlib import Path

# Allow importing clfkit from the repo root when running this notebook
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd
from sklearn.datasets import make_classification

from clfkit import benchmark
```

```python
X, y = make_classification(
    n_samples=400,
    n_features=12,
    n_informative=8,
    n_redundant=2,
    n_classes=2,
    random_state=42,
)

df = pd.DataFrame(X, columns=[f"f{i}" for i in range(X.shape[1])])
df["label"] = y
df.head()
```

```python
# Smaller model set + fewer CV folds for a fast demo.
# In the contest, drop `models=` to run the full GridSearch suite.
result = benchmark(
    df,
    target="label",
    cv=3,
    n_jobs=-1,
    models=[
        "LogisticRegression",
        "RandomForestClassifier",
        "XGBClassifier",
        "LGBMClassifier",
        "CatBoostClassifier",
    ],
)

result
```

```python
print("best_model :", result.best_model)
print("best_score :", result.best_score)
print("best_params:", result.best_params)
result.leaderboard
```
